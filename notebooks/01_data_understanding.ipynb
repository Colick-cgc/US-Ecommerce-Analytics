# 数据理解与数据质量审计
## 1.1 八张表的结构梳理
|数据表|数据粒度|主要用途|
|---|---|---|
|customers|每个订单对应的客户记录|客户画像、客户分层|
|geolocation|每个邮编前缀的地理记录|区域分析|
|orders|每笔订单|订单趋势、订单状态|
|order_items|每笔订单中的一个商品明细|商品销售、利润|
|order_payments	|每笔订单的一条支付记录|支付方式、支付金额|
|order_reviews|每条评价记录|客户满意度|
|products|每个商品|品类、品牌、成本|
|sellers|每个卖家|卖家维度分析|
## 1.2 Pandas 数据质量检查
|检查项目	|Pandas方法|	检查目的|
|---|---|---|
|数据规模|	shape|	行数、列数|
|字段类型|	dtypes、info()|	是否存在错误类型|
|缺失值|	isnull().sum()|	是否有缺失数据|
|重复记录|	duplicated()|	是否存在重复|
|唯一值数量|	nunique()|	判断字段基数|
|数值分布|	describe()|	检查异常范围|
|分类分布|	value_counts()|	检查类别是否合理|

## 1.3 八张表的相互连接

|关系|连接字段|说明|
|---|---|---|
|customer--orders|customer_id|文件说 customer_id 是订单级一次性标识；customer_unique_id 才是客户级分析键|
|orders--order_items|order_id|一个订单多个商品项|
|orders--order_payments|order_id|有 payment_sequential=2，说明可多笔支付|
|orders--order_reviews|order_id|实际可验证一个订单是否只有一条评论|
|products--order_items|product_id|一个商品可出现在多个订单项|
|sellers--order_items|seller_id|一个卖家可卖多个订单项|
|geolocation--customer|zip_code_prefix = customer_zip_code_prefix|一个邮编前缀可对应多个客户|
|geolocation--sellers|zip_code_prefix = seller_zip_code_prefix|一个邮编前缀可对应多个卖家|
